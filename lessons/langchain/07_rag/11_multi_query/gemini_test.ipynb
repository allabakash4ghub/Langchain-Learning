```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os
from langchain_core.messages import SystemMessage,HumanMessage
```

```python
load_dotenv()

llm=ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0.1,
    max_tokens=400    
)

response=llm.invoke("can you write four sentence about langchain and each line in new line")
print(f"type of response : {type(response)}")
print(f"type of response of content: {type(response.content)}")
print(response)
print("*"*30)

response.content[0]["text"]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
# Initialize the free text embedding model
embedding=GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001"
)
```

```python
# Generate a vector representation of your text
text_vector =embedding.embed_query("How do I use embeddings in a RAG system?")
print(f"Generated an embedding with {len(text_vector)} dimensions!")
```

## Cosine similarity

Cosine similarity measures the orientation between two non-zero vectors in a high-dimensional space by calculating the cosine of the angle (θ) between them. It focuses on direction rather than magnitude
Cosine similarity is a metric used to measure how similar the documents are irrespective of their size. It is the cosine of the angle between two vectors.

The cosine similarity is advantageous because even if the two similar documents are far apart by the Euclidean distance (due to the size of the document), chances are they may still be oriented closer together. The smaller the angle, higher the cosine similarity.

A commonly used approach to match similar documents is based on counting the maximum number of common words between the documents.

But this approach has an inherent flaw.

That is, as the size of the document increases, the number of common words tend to increase even if the documents talk about different topics. The cosine similarity helps overcome this fundamental flaw in the ‘count-the-common-words’ or Euclidean distance approach.

Mathematically, Cosine similarity measures the cosine of the angle between two vectors projected in a multi-dimensional space.

As a similarity metric, how does cosine similarity differ from the number of common words?

When plotted on a multi-dimensional space, where each dimension corresponds to a word in the document, the cosine similarity captures the orientation (the angle) of the documents and not the magnitude. If you want the magnitude, compute the Euclidean distance instead.

Score Range & Meaning

• 1: The vectors point in the exact same direction (θ = 0°), meaning maximum similarity.  
• 0: The vectors are orthogonal or perpendicular (θ = 90°), meaning no directional relationship.   \
• -1: The vectors point in opposite directions (θ = 180°), meaning complete dissimilarity.




Common Applications

• Natural Language Processing (NLP): Comparing text documents or word embeddings by turning word frequencies into high-dimensional vectors.  
• Recommendation Systems: Measuring user preferences or item ratings to find similar profiles.  
• Information Retrieval: Ranking search results based on semantic relevance

```python
def cosine_similarity(vec1, vec2):
    """
    Calculate the cosine similarity between two vectors.

    Args:
        vec1 (list): First vector.
        vec2 (list): Second vector.

    Returns:
        float: Cosine similarity between vec1 and vec2.
    """
    dot_product = sum(a * b for a, b in zip(vec1, vec2))
    magnitude_vec1 = sum(a ** 2 for a in vec1) ** 0.5
    magnitude_vec2 = sum(b ** 2 for b in vec2) ** 0.5

    if magnitude_vec1 == 0 or magnitude_vec2 == 0:
        return 0.0

    return dot_product / (magnitude_vec1 * magnitude_vec2)
```

```python
import math
def cosine_similarity2(vec1,vec2):
    """ calculate consine similarity between two vectors
    
    Args:
        vec1 (list): First vector.
        vec2 (list): Second vector.
        
    Returns:
            float: Cosine similarity between vec1 and vec2."""

    dot_product = sum(a*b for a,b in zip(vec1,vec2))
    magnitue_vec1 = math.sqrt(sum(a**2 for a in vec1))
    magnitue_vec2 = math.sqrt(sum(b**2 for b in vec2))
    
    if magnitue_vec1 == 0 or magnitue_vec2 == 0:
        return 0
    
    return dot_product / (magnitue_vec1 * magnitue_vec2)
```

```python
vec1 = [1, 2, 3]
vec2 = [9, 9, 8]
cosine_similarity2(vec1, vec2)
```

### For production applications, particularly Data Science, Machine Learning and GenAI applications, I would recommend a third implementation using NumPy.

```python
import numpy as np

def cosine_similarity3(vec1, vec2):
    vec1 = np.asarray(vec1, dtype=float)
    vec2 = np.asarray(vec2, dtype=float)

    if vec1.shape != vec2.shape:
        raise ValueError("Vectors must have the same shape")

    magnitude_vec1 = np.linalg.norm(vec1)
    magnitude_vec2 = np.linalg.norm(vec2)

    if magnitude_vec1 == 0 or magnitude_vec2 == 0:
        return 0.0

    dot_product = np.dot(vec1, vec2)

    return float(dot_product / (magnitude_vec1 * magnitude_vec2))
```

```python
vec1 = [1, 2, 3]
vec2 = [9, 9, 8]
cosine_similarity3(vec1, vec2)
```
